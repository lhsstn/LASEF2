```python
import json

data1 = []
with open("/home/work/mlp/hslim/LASEF2/data/results/SkelLang/translated_skeleton/Qwen2.5-7B-Instruct-skeleton_multiturn_skelLang-ko.jsonl", "r", encoding="utf-8") as f:
    for line in f:
        data1.append(json.loads(line))
```

```python
import json

data2 = []
with open("/home/work/mlp/hslim/LASEF2/data/results/SkelLang/translated_skeleton/Qwen2.5-7B-Instruct-skeleton_multiturn_skelLang-en.jsonl", "r", encoding="utf-8") as f:
    for line in f:
        data2.append(json.loads(line))
```

```python
i = 0
```

```python
from IPython.display import display, Markdown

def compare_samples(idx):
    # 인덱스 범위 초과 방지
    if idx >= len(data1) or idx >= len(data2):
        print("❌ 인덱스가 데이터 범위를 초과했습니다.")
        return
        
    prompt = data1[idx].get('prompt', 'N/A')
    q_lang = data1[idx].get('question_language', 'N/A')
    gold_ans = data1[idx].get('answer', 'N/A')
    
    skel1 = data1[idx]['skeleton'][0]
    resp1 = data1[idx]['responses'][0]
    
    skel2 = data2[idx]['skeleton'][0]
    resp2 = data2[idx]['responses'][0]

    # Markdown 양식 조립
    md_content = f"""
## 🔍 Sample #{idx} 비교 (Global ID: {data1[idx].get('global_id', idx)})
* **질문 언어:** `{q_lang}` | **실제 정답 (Gold):** `{gold_ans}`

### ❓ 원본 질문 (Question)
> {prompt}

---

### 🔴 데이터 1 (Data 1)
* **Reasoning Skeleton:**
{skel1}

* **모델 답변 (Response):**
{resp1}

---

### 🔵 데이터 2 (Data 2)
* **Reasoning Skeleton:**
{skel2}

* **모델 답변 (Response):**
{resp2}

<br>
<div style="border-top: 3px double #bbb; margin: 20px 0;"></div>
"""
    
    # 렌더링 출력
    display(Markdown(md_content))
```

```python
compare_samples(i)
i += 1
```
