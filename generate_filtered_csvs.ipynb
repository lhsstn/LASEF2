# Generate Filtered CSVs by Excluding Languages

This notebook recursively scans the `dataframe/` directory in the project path and automatically creates a filtered version (`_filtered.csv`) for any existing unfiltered strict/pure delta/count CSV files by dropping the specified languages.

```python
import os
import glob
import pandas as pd

PRJ_PATH = "/home/work/mlp/hslim/LASEF2"
EXCLUDE_LANGS = ["kn", "ku", "jv", "mg", "so", "gn", "uz", "su", "yo", "qu", "ceb", "ka", "tg", "my", "km"]

print(f"🚀 Project Path: {PRJ_PATH}")
print(f"🚫 Exclude Languages ({len(EXCLUDE_LANGS)} total): {EXCLUDE_LANGS}")
```

```python
dataframe_dir = os.path.join(PRJ_PATH, "dataframe")
if not os.path.exists(dataframe_dir):
    print(f"❌ Error: {dataframe_dir} directory does not exist.")
else:
    # Recursively find all CSV files under dataframe/
    all_csvs = glob.glob(os.path.join(dataframe_dir, "**", "*.csv"), recursive=True)
    unfiltered_csvs = [f for f in all_csvs if not f.endswith("_filtered.csv")]
    
    print(f"📂 Found {len(all_csvs)} CSV files total.")
    print(f"🔍 Found {len(unfiltered_csvs)} unfiltered CSV files to process.")
    print("-" * 80)
    
    processed_count = 0
    for csv_path in unfiltered_csvs:
        dir_name = os.path.dirname(csv_path)
        base_name = os.path.basename(csv_path)
        name_part, ext_part = os.path.splitext(base_name)
        
        # Construct target filtered file path
        filtered_path = os.path.join(dir_name, f"{name_part}_filtered.csv")
        
        # Load original CSV
        df = pd.read_csv(csv_path, index_col=0)
        
        # Filter out exclude languages cleanly from both index and columns (if applicable)
        df_filtered = df.drop(index=EXCLUDE_LANGS, errors='ignore')
        df_filtered = df_filtered.drop(columns=[col for col in EXCLUDE_LANGS if col in df_filtered.columns], errors='ignore')
        
        # Save as filtered CSV
        df_filtered.to_csv(filtered_path, index=True)
        processed_count += 1
        print(f"✅ [{processed_count}] Filtered & Saved:\n   • From: {csv_path}\n   • To  : {filtered_path}\n")
        
    print("-" * 80)
    print(f"🎉 Successfully processed {processed_count} CSV files and generated their filtered versions!")
```
